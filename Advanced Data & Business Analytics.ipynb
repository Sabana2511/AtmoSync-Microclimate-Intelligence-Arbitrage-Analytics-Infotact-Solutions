# **3. Data Transformation**

In this phase, raw variables are transformed into analytical features, datetime components, logarithmic scales to handle skewed distributions, and categorical risk flags for business monitoring.

```python
import pandas as pd
import numpy as np

# 1. Load the cleaned dataset
df = pd.read_csv('Euro_Crops.csv')

# Handle any remaining infinity representations
df = df.replace([np.inf, -np.inf], np.nan)
print(f"Dataset Loaded Successfully! Shape: {df.shape}\n")

# 2. Datetime Transformations & Temporal Features
df['Timestamp'] = pd.to_datetime(df['Unnamed: 0'], errors='coerce')
df['Harvest_Date'] = pd.to_datetime(df['Harvest_Date'], errors='coerce')

df['Year'] = df['Timestamp'].dt.year
df['Month'] = df['Timestamp'].dt.month
df['Month_Name'] = df['Timestamp'].dt.strftime('%b')
df['Day_Of_Week'] = df['Timestamp'].dt.day_name()
df['Is_Weekend'] = df['Timestamp'].dt.dayofweek.isin([5, 6]).astype(int)

# Calculate duration between harvesting and transit logging
df['Days_In_Transit'] = (df['Timestamp'] - df['Harvest_Date']).dt.days
print("================================================================================")
print("DATETIME PARSING & TEMPORAL FEATURE EXTRACTION")
print("================================================================================")
print(df[['Timestamp', 'Harvest_Date', 'Days_In_Transit', 'Year', 'Month_Name', 'Day_Of_Week', 'Is_Weekend']].head())

print("\nYear Distribution:\n", df['Year'].value_counts().sort_index())
print("\nWeekday vs Weekend Trip Counts:\n", df['Is_Weekend'].value_counts().rename({0: 'Weekday', 1: 'Weekend'}))

# 3. Logarithmic Scaling & Outlier Capping
# Stabilize heavily right-skewed variables for statistical modeling
skewed_cols = ['Delivery_Time', 'Vibration_Level', 'Fuel_Costs', 'Storage_Temperature']

for col in skewed_cols:
    p01 = df[col].quantile(0.01)
    p99 = df[col].quantile(0.99)
    df[f'{col}_Capped'] = np.clip(df[col], p01, p99)
    df[f'Log_{col}'] = np.log1p(df[f'{col}_Capped'])
print("\n================================================================================")
print("OUTLIER CAPPING & LOGARITHMIC TRANSFORMATIONS")
print("================================================================================")
print(df[['Delivery_Time', 'Delivery_Time_Capped', 'Log_Delivery_Time', 'Fuel_Costs', 'Fuel_Costs_Capped', 'Log_Fuel_Costs']].head())

# Comparison table between Raw, Capped, and Log transformed stats
stats_comparison = pd.DataFrame({
    'Raw Mean': [df['Delivery_Time'].mean(), df['Fuel_Costs'].mean()],
    'Capped Mean': [df['Delivery_Time_Capped'].mean(), df['Fuel_Costs_Capped'].mean()],
    'Log Mean': [df['Log_Delivery_Time'].mean(), df['Log_Fuel_Costs'].mean()],
    'Raw Max': [df['Delivery_Time'].max(), df['Fuel_Costs'].max()],
    'Capped Max': [df['Delivery_Time_Capped'].max(), df['Fuel_Costs_Capped'].max()],
    'Log Max': [df['Log_Delivery_Time'].max(), df['Log_Fuel_Costs'].max()]
}, index=['Delivery_Time', 'Fuel_Costs'])

print("\nComparison Table (Raw vs Capped vs Log Stats):")
print(stats_comparison.round(2))

# 4. Categorical Binning & Risk Flags
# Threshold-based binary flags for high-risk conditions
df['High_Spoilage_Risk_Flag'] = (df['Spoilage_Risk'] > 1.20).astype(int)
df['High_Queue_Delay_Flag'] = (df['Queue_Time'] > 5.0).astype(int)

# Transit duration classification into tertiles
df['Transit_Duration_Class'] = pd.qcut(df['Delivery_Time_Capped'], q=3, labels=['Short', 'Medium', 'Long'])
print("\n================================================================================")
print("BUSINESS RISK FLAGS & CATEGORICAL BINNING")
print("================================================================================")
print(df[['Spoilage_Risk', 'High_Spoilage_Risk_Flag', 'Queue_Time', 'High_Queue_Delay_Flag', 'Transit_Duration_Class']].head())

print("\n1. High Spoilage Risk Flag (%):")
print((df['High_Spoilage_Risk_Flag'].value_counts(normalize=True) * 100).round(2))

print("\n2. High Queue Delay Flag (%):")
print((df['High_Queue_Delay_Flag'].value_counts(normalize=True) * 100).round(2))

print("\n3. Transit Duration Classification Counts:")
print(df['Transit_Duration_Class'].value_counts())

# 5. Export Transformed Dataset
df.to_csv('EURO_Crops_Transformed.csv', index=False)
print("Data Transformation complete. File saved as 'EURO_Crops_Transformed.csv'")
```

# **5. Advanced Data & Business Analytics**

**5.1. Cohort & Temporal Trend Analysis**

Below are the aggregated metrics tracking seasonality and dispatch cycles across months and weekday/weekend operational periods.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Load transformed dataset
df = pd.read_csv('EURO_Crops_Transformed.csv')

# 1. Monthly Trends Execution
df['Month_Num'] = pd.to_datetime(df['Timestamp']).dt.month
monthly_trends = (
    df.groupby(['Month_Num', 'Month_Name'])
    .agg(
        Trip_Count=('Spoilage_Risk', 'count'),
        Avg_Spoilage_Risk=('Spoilage_Risk', 'mean'),
        Avg_Queue_Time=('Queue_Time', 'mean'),
        Avg_Fuel_Costs=('Fuel_Costs_Capped', 'mean'),
        High_Spoilage_Rate=('High_Spoilage_Risk_Flag', 'mean'),
    )
    .reset_index()
    .sort_values('Month_Num')
)

# 2. Visualizations
sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=monthly_trends,
    x='Month_Name',
    y='High_Spoilage_Rate',
    palette='Blues_d',
    ax=axes[0],
)
axes[0].set_title(
    'High Spoilage Risk Rate (>1.20) by Month', fontsize=12, fontweight='bold'
)
axes[0].set_ylabel('Proportion of High Risk Trips')

weekend_trends = (
    df.groupby('Is_Weekend')
    .agg(
        Avg_Queue_Time=('Queue_Time', 'mean'),
    )
    .reset_index()
)
weekend_trends['Dispatch_Type'] = weekend_trends['Is_Weekend'].map(
    {0: 'Weekday', 1: 'Weekend'}
)

sns.barplot(
    data=weekend_trends,
    x='Dispatch_Type',
    y='Avg_Queue_Time',
    palette='Set2',
    ax=axes[1],
)
axes[1].set_title(
    'Average Queue Delay (Hours): Weekday vs Weekend',
    fontsize=12,
    fontweight='bold',
)

plt.tight_layout()
plt.show()
```

**January Spoilage Peak**:*January exhibits the highest high-spoilage occurrence rate at 30.62%, compared to an annual baseline average of 29.13%.*   

**December Queue Bottlenecks**: *Average queue delays peak in December at 5.30 hours, driven by end-of-year shipping volume surges.*  

**Uniform Weekend Friction:** *Queue delay occurrences remain virtually identical between weekdays (38.06%) and weekends (38.24%), confirming that delays stem from facility capacity bottlenecks rather than weekday traffic congestion.*

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load dataset
df = pd.read_csv('EURO_Crops_Transformed.csv')

# Aggregation logic
bottleneck_summary = (
    df.groupby(['Vehicle_Type', 'Transit_Duration_Class'], observed=False)
    .agg(
        Trip_Count=('Spoilage_Risk', 'count'),
        Avg_Queue_Time=('Queue_Time', 'mean'),
        High_Queue_Delay_Rate=('High_Queue_Delay_Flag', 'mean'),
        Avg_Fuel_Costs=('Fuel_Costs_Capped', 'mean'),
        Avg_Spoilage_Risk=('Spoilage_Risk', 'mean'),
    )
    .reset_index()
)

# Visualizations
sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=bottleneck_summary,
    x='Vehicle_Type',
    y='High_Queue_Delay_Rate',
    hue='Transit_Duration_Class',
    palette='Blues_d',
    ax=axes[0],
)
axes[0].set_title(
    'Queue Delay Rate (>5 Hrs) by Vehicle & Duration Class',
    fontsize=12,
    fontweight='bold',
)
axes[0].set_ylabel('Proportion Delayed')

sns.barplot(
    data=bottleneck_summary,
    x='Vehicle_Type',
    y='Avg_Fuel_Costs',
    hue='Transit_Duration_Class',
    palette='YlOrRd',
    ax=axes[1],
)
axes[1].set_title(
    'Average Capped Fuel Costs ($) by Transport Mode',
    fontsize=12,
    fontweight='bold',
)
axes[1].set_ylabel('Mean Fuel Cost ($)')

plt.tight_layout()
plt.show()
```

**5.3. Quality Degradation & Financial Impact Analysis**

Mapping queue time severity against crop quality maintenance and operational expenditure reveals the direct impact of logistics delays.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load dataset
df = pd.read_csv('EURO_Crops_Transformed.csv')

# Queue time binning
df['Queue_Time_Bin'] = pd.cut(
    df['Queue_Time'],
    bins=[0, 2, 4, 6, 8, 10, float('inf')],
    labels=['0-2h', '2-4h', '4-6h', '6-8h', '8-10h', '>10h'],
)

# Aggregation logic
impact_summary = (
    df.groupby('Queue_Time_Bin', observed=False)
    .agg(
        Trip_Count=('Spoilage_Risk', 'count'),
        Avg_Quality_Ratio=('Quality_Maintenance_Ratio', 'mean'),
        Avg_Fuel_Costs=('Fuel_Costs_Capped', 'mean'),
    )
    .reset_index()
)

# Visualization
sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.lineplot(
    data=impact_summary,
    x='Queue_Time_Bin',
    y='Avg_Quality_Ratio',
    marker='o',
    color='crimson',
    linewidth=2.5,
    ax=axes[0],
)
axes[0].set_title(
    'Quality Maintenance Ratio Drop across Queue Time Bins',
    fontsize=12,
    fontweight='bold',
)
axes[0].set_ylabel('Mean Quality Maintenance Ratio')

crop_degradation = (
    df.groupby(['Crop_Type', 'High_Queue_Delay_Flag'])
    .agg(Avg_Quality_Ratio=('Quality_Maintenance_Ratio', 'mean'))
    .reset_index()
)

sns.barplot(
    data=crop_degradation,
    x='Crop_Type',
    y='Avg_Quality_Ratio',
    hue='High_Queue_Delay_Flag',
    palette='Set1',
    ax=axes[1],
)
axes[1].set_title(
    'Quality Ratio by Crop: Normal Queue (0) vs High Delay (1)',
    fontsize=12,
    fontweight='bold',
)

plt.tight_layout()
plt.show()
```

**Exponential Quality Decay**: *Quality Maintenance Ratio drops precipitously from 214.31 in trips under 2 hours down to 19.83 when queue delays exceed 10 hours—a 90.7% total decay in quality retention.*   

**Wheat Sensitivity**: *Wheat suffers the highest relative quality degradation under extreme delays, dropping 71.22% in quality retention when queue time exceeds 5 hours.*   

**Idle Fuel Surcharge**: *Vehicles trapped in the 8–10 hour queue bin record the highest average fuel expenditure ($182.31), reflecting continuous idle engine runtime during severe facility delays.*
